```python
import sys
# arch is not the easiest package to install, but its more stable and full featured than tsbootstrap.
# the manual is at https://app.readthedocs.org/projects/arch/downloads/pdf/stable/
from arch.bootstrap import StationaryBootstrap
```

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
```

```python
data = np.random.random(252).reshape(63,4) # 63 observations in a quarter
```

```python
def get_eigvec(a):
    m = a.T@a
    return np.linalg.eigh(m)[1][-1]
```

```python
sb = StationaryBootstrap(8,data)
```

```python
results = sb.apply(get_eigvec,1000)
```

```python
df = pd.DataFrame(results,columns = 'x y z w'.split())
```

```python
ax = sns.histplot(df,bins=100,kde=True,alpha=0.25,stat='percent')
ax.grid(True)
```
